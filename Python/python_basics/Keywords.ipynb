# Keywords 

## A Keywords in python or any programming language is a Special meaning and reserved words that not be used as a variable , class , function name or they have a reserved meaning in python

###
| Keyword    | Purpose                           |
| ---------- | --------------------------------- |
| `False`    | Boolean false value               |
| `None`     | Represents no value               |
| `True`     | Boolean true value                |
| `and`      | Logical AND                       |
| `as`       | Creates an alias                  |
| `assert`   | Debugging check                   |
| `async`    | Asynchronous programming          |
| `await`    | Waits for an async task           |
| `break`    | Exits a loop                      |
| `class`    | Defines a class                   |
| `continue` | Skips current loop iteration      |
| `def`      | Defines a function                |
| `del`      | Deletes an object                 |
| `elif`     | Else if condition                 |
| `else`     | Alternative condition             |
| `except`   | Handles exceptions                |
| `finally`  | Always executes after `try`       |
| `for`      | Creates a loop                    |
| `from`     | Imports specific modules          |
| `global`   | Declares a global variable        |
| `if`       | Conditional statement             |
| `import`   | Imports a module                  |
| `in`       | Checks membership                 |
| `is`       | Checks object identity            |
| `lambda`   | Creates an anonymous function     |
| `nonlocal` | Refers to outer function variable |
| `not`      | Logical NOT                       |
| `or`       | Logical OR                        |
| `pass`     | Empty statement                   |
| `raise`    | Raises an exception               |
| `return`   | Returns a value from a function   |
| `try`      | Starts exception handling         |
| `while`    | Creates a while loop              |
| `with`     | Simplifies resource management    |
| `yield`    | Returns values from a generator   |
| `match`    | Pattern matching (Python 3.10+)   |
| `case`     | Used with `match`                 |



```python
import keyword

print(keyword.iskeyword("for"))
print(keyword.iskeyword("Apple"))
```
