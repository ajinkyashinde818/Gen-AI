# Task 2: Categories (Sets)

```python
# 1. From your products list, create a set of categories called categories_set.

categories = ["Electronics", "Accessories", "Accessories", "Electronics", "Accessories"]

categories_set = set(categories)

print(categories_set)
```

```python
# 2. Add a new category and show that duplicates are ignored.

categories_set.add("Office")

print("After Adding Office:", categories_set)

categories_set.add("Office")

print("After Adding Office Again:", categories_set)
```

```python
# 3. Check whether a category exists in the set.

print("Office" in categories_set)
print("Gaming" in categories_set)
```

```python
# Extra (optional): Print the total number of unique categories.

print("Total Categories:", len(categories_set))
```

# Task 3: Product Pricing (Dictionaries)

```python
# 1. Create a dictionary named price_dict.

price_dict = {
    "Laptop": 70000,
    "Wireless Mouse": 800,
    "Mechanical Keyboard": 2500,
    "Monitor": 15000,
    "USB-C Cable": 500,
    "Docking Station": 3000
}

print(price_dict)
```

```python
# 2. Add a new product with price.

price_dict["Ear Buds"] = 2500

print(price_dict)
```

```python
# 3. Update the price of an existing product.

price_dict["Laptop"] = 65000

print(price_dict)
```

```python
# 4. Remove a product.

del price_dict["USB-C Cable"]

print(price_dict)
```

```python
# 5. Print the average price.

total = 0

for price in price_dict.values():
    total = total + price

average = total / len(price_dict)

print("Average Price:", average)
```

```python
# Extra (optional): Print the maximum and minimum price.

print("Maximum Price:", max(price_dict.values()))
print("Minimum Price:", min(price_dict.values()))
```
