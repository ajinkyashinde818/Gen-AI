# What is a Module?

A **module** is a single Python file (`.py`) containing reusable Python code.

It can contain:
- Functions
- Variables
- Classes

📄 **Module = One Python file**


````python
# calculator.py

def add(a, b):
    return a + b

def sub(a, b):
    return a - b
````

## Using the Module

````python
import calculator

print(calculator.add(10, 5))
print(calculator.sub(10, 5))
````

# What is a Package?

A **package** is a collection of related Python modules organized together.

📁 **Package = Folder containing modules**

Example:

```text
MathPackage/
├── calculator.py
├── geometry.py
└── statistics.py
```


# What is a Library?

A **library** is a larger collection of reusable code that can contain packages and modules.

Examples:
- NumPy
- Pandas
- Matplotlib
- TensorFlow

📚 **Library = Collection of reusable packages and modules**


# Easy Analogy

| Real World | Python |
|---|---|
| 📄 One Book | Module |
| 📚 Bookshelf | Package |
| 🏛️ Library Building | Library |


# Relationship

```text
Library
│
├── Package
│   ├── Module
│   └── Module
│
└── Package
    ├── Module
    └── Module
```

### Remember:

**Module → Package → Library**

# Module vs Package vs Library

| Module | Package | Library |
|---|---|---|
| Single `.py` file | Collection of modules | Larger reusable collection |
| Smallest unit | Organizes related modules | Can contain packages/modules |
| Example: a Python file | Folder of Python modules | NumPy, Pandas |


# Why Do We Use Them?

- Code reuse
- Better organization
- Easier maintenance
- Large projects become manageable
- Multiple developers can work on different modules


# Creating Your Own Module

### Step 1: Create `my_module.py`

```python
def greet(name):
    return "Hello " + name

def square(number):
    return number * number
```

### Step 2: Use it in another file

```python
import my_module

print(my_module.greet("Dadus"))
print(my_module.square(5))
```
