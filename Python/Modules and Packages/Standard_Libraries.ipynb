# 1. math Module

### What is `math`?

The `math` module provides mathematical functions and constants.

### Why do we use it?

Python has basic operators like `+`, `-`, `*`, and `/`, but the `math` module provides more advanced mathematical operations.

### Important functions

- `math.sqrt(x)` – square root
- `math.ceil(x)` – rounds upward
- `math.floor(x)` – rounds downward
- `math.factorial(x)` – factorial
- `math.pow(x, y)` – power
- `math.pi` – value of π


### Where is it useful?

Useful in mathematics programs, scientific programs, games, engineering, and calculations.

```python
import math

print(math.sqrt(25))
print(math.ceil(4.2))
print(math.floor(4.8))
print(math.factorial(5))
print(math.pi)
```

# 2. random Module

### What is `random`?

The `random` module is used to generate random values.

### Why do we use it?

It is useful when we want Python to select or generate values unpredictably.

### Important functions

- `random.randint(a, b)` – random integer between a and b
- `random.choice(sequence)` – selects one random item
- `random.shuffle(list)` – changes the order of list items randomly
- `random.random()` – random decimal between 0 and 1


### Where is it useful?

Useful in games, quizzes, simulations, and random selection.

```python
import random

print(random.randint(1, 10))

names = ["Amit", "Rahul", "Priya"]
print(random.choice(names))
```

# 3. datetime Module

### What is `datetime`?

The `datetime` module is used to work with dates and time.

### Important concepts

- `date` – works with dates
- `time` – works with time
- `datetime` – works with both date and time
- `timedelta` – represents a difference between dates or times


### Where is it useful?

Useful in attendance systems, booking systems, reminders, and timestamps.

```python
from datetime import datetime, date

print(date.today())
print(datetime.now())
```

# 4. os Module

### What is `os`?

The `os` module allows a Python program to interact with the operating system.

### Important functions

- `os.getcwd()` – get current directory
- `os.chdir()` – change directory
- `os.listdir()` – list files and folders
- `os.mkdir()` – create a folder
- `os.remove()` – delete a file
- `os.rename()` – rename a file or folder

You have already studied this module in detail.


### Where is it useful?

Useful for automation, file management, and system-related programs.

```python
import os

print(os.getcwd())
print(os.listdir())
```

# 5. sys Module

### What is `sys`?

The `sys` module provides information and functions related to the Python interpreter and runtime environment.

### Important uses

- Check Python version
- Check platform
- Access command-line arguments
- Work with interpreter-related information

### Important attributes

- `sys.version`
- `sys.platform`
- `sys.argv`


### Where is it useful?

Useful in command-line applications and programs that need Python runtime information.

```python
import sys

print(sys.version)
print(sys.platform)
```

# 6. pathlib Module

### What is `pathlib`?

The `pathlib` module provides a modern object-oriented way to work with file and folder paths.

### Important functions and methods

- `Path.cwd()` – current directory
- `Path.exists()` – check whether a path exists
- `Path.name` – get file name
- `Path.suffix` – get file extension
- `Path.glob()` – search for files

`pathlib` is often cleaner and easier than manually joining path strings.


### Where is it useful?

Useful for modern file and folder handling.

```python
from pathlib import Path

current = Path.cwd()
print(current)

file = Path("student.txt")
print(file.exists())
```

# 7. json Module

### What is `json`?

The `json` module is used to work with JSON data.

JSON is a common format for storing and exchanging structured data.

### Important functions

- `json.dumps()` – Python object to JSON string
- `json.loads()` – JSON string to Python object
- `json.dump()` – write JSON to a file
- `json.load()` – read JSON from a file


### Where is it useful?

Useful in APIs, configuration files, web applications, and data storage.

```python
import json

student = {"name": "Dadus", "marks": 90}

json_text = json.dumps(student)
print(json_text)

python_data = json.loads(json_text)
print(python_data)
```

# 8. csv Module

### What is `csv`?

The `csv` module is used to read and write CSV files.

CSV means Comma-Separated Values.

A CSV file is commonly used for tabular data such as student records and marks.

### Important tools

- `csv.reader()` – read CSV data
- `csv.writer()` – write CSV data
- `csv.DictReader()` – read rows as dictionaries
- `csv.DictWriter()` – write dictionaries


### Where is it useful?

Useful for student records, reports, spreadsheets, and data analysis.

```python
import csv

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Marks"])
    writer.writerow(["Rahul", 85])
```

# 9. collections Module

### What is `collections`?

The `collections` module provides specialized data structures.

### Important tools

- `Counter` – counts occurrences
- `deque` – efficient queue-like structure
- `defaultdict` – dictionary with default values

These structures are useful when normal lists and dictionaries are not the best choice.


### Where is it useful?

Very useful in DSA, counting problems, queues, and efficient data handling.

```python
from collections import Counter, deque

print(Counter("programming"))

queue = deque([1, 2, 3])
queue.append(4)
queue.popleft()

print(queue)
```

# 10. statistics Module

### What is `statistics`?

The `statistics` module provides functions for basic statistical calculations.

### Important functions

- `mean()` – average
- `median()` – middle value
- `mode()` – most frequently occurring value


### Where is it useful?

Useful for marks analysis, reports, and data analysis.

```python
import statistics

numbers = [70, 80, 90, 100]

print(statistics.mean(numbers))
print(statistics.median(numbers))
```

# 11. re Module

### What is `re`?

The `re` module is used for Regular Expressions.

Regular Expressions help us search, match, and extract patterns from text.

### Important functions

- `re.findall()` – find all matching patterns
- `re.search()` – find the first match
- `re.match()` – check a match at the beginning
- `re.sub()` – replace matching text


### Where is it useful?

Useful for email validation, phone number extraction, searching text, and text processing.

```python
import re

text = "My marks are 85 and 90"

numbers = re.findall(r"\d+", text)
print(numbers)
```

Be careful with delete operations because they can permanently remove files.

# 13. logging Module

### What is `logging`?

The `logging` module records events that happen while a program is running.

### Logging levels

- DEBUG
- INFO
- WARNING
- ERROR
- CRITICAL

Logging is better than using many `print()` statements in professional applications.


### Where is it useful?

Useful for debugging, monitoring, and professional applications.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Program started")
logging.warning("This is a warning")
```

# 14. sqlite3 Module

### What is `sqlite3`?

The `sqlite3` module provides a lightweight SQL database that is included with Python.

### Basic process

1. Connect to database
2. Create a table
3. Insert data
4. Read data
5. Update or delete data
6. Close the connection


### Where is it useful?

Useful for student management systems, small applications, and local databases.

```python
import sqlite3

connection = sqlite3.connect("college.db")

print("Database connected")

connection.close()
```
