# Exception Handling in Python
## Complete Notes, Examples and Practice Questions

# 1. What is Exception Handling?

An **exception** is an error that happens while a program is running.

**Exception handling** allows us to handle these errors properly so the program does not stop unexpectedly.

Example: dividing by zero causes `ZeroDivisionError`.

````python
print(10 / 0)
````

# 2. Why Do We Need Exception Handling?

Exception handling helps us:

- Handle invalid user input
- Handle division by zero
- Handle missing files
- Prevent unexpected program crashes
- Display useful messages to users

The basic idea is:

**Try risky code → If an error happens → Handle it properly**

# 3. try and except

Basic syntax:

```python
try:
    # Code that may cause an error
except:
    # Code that handles the error
```

````python
try:
    print(10 / 0)
except:
    print("Something went wrong!")
````

# 4. Specific Exceptions

It is better to handle the specific error when possible.

Example: `ZeroDivisionError`

````python
try:
    num1 = 10
    num2 = 0
    print(num1 / num2)
except ZeroDivisionError:
    print("You cannot divide by zero.")
````

# 5. ValueError

A `ValueError` commonly happens when the user enters an invalid value.

For example, converting `"hello"` into an integer.

````python
try:
    age = int(input("Enter your age: "))
    print("Your age is:", age)
except ValueError:
    print("Please enter a valid number.")
````

# 6. Multiple except Blocks

A program can have multiple possible errors. We can handle them separately.

````python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except ValueError:
    print("Please enter a valid integer.")
except ZeroDivisionError:
    print("You cannot divide by zero.")
````

# 7. Common Exceptions

| Exception | Meaning |
|---|---|
| `ValueError` | Invalid value |
| `TypeError` | Invalid operation between data types |
| `ZeroDivisionError` | Division by zero |
| `NameError` | Variable or name does not exist |
| `IndexError` | Invalid list index |
| `KeyError` | Dictionary key does not exist |
| `FileNotFoundError` | File does not exist |

# 8. else Block

The `else` block runs only when **no exception occurs**.

```python
try:
    # risky code
except:
    # runs when an error occurs
else:
    # runs when no error occurs
```

````python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input!")
else:
    print("You entered:", number)
````

# 9. finally Block

The `finally` block **always runs**, whether an error occurs or not.

It is useful for cleanup operations such as closing files or database connections.

````python
try:
    number = int(input("Enter a number: "))
    print(number)
except ValueError:
    print("Invalid input!")
finally:
    print("Program finished.")
````

# 10. Complete Structure

```text
try → except → else → finally
```

- `try` → Code that may cause an error
- `except` → Handles the error
- `else` → Runs when there is no error
- `finally` → Always runs

````python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
except ValueError:
    print("Please enter a valid integer.")
except ZeroDivisionError:
    print("You cannot divide by zero.")
else:
    print("Result:", result)
finally:
    print("Program execution finished.")
````

# 11. Exception as a Variable

You can store the error message using `as`.

Syntax:

```python
except Exception as error:
    print(error)
```

````python
try:
    number = int("hello")
except ValueError as error:
    print("Error:", error)
````

# 12. raise

The `raise` keyword allows us to intentionally create an exception when a program rule is broken.

````python
age = int(input("Enter your age: "))

if age < 0:
    raise ValueError("Age cannot be negative.")

print("Age:", age)
````

# 13. Important Rule

Avoid using a general `except:` when you know the expected error.

Better:

```python
except ValueError:
    print("Invalid number")
```

Specific exception handling makes programs easier to understand and debug.

# 14. Real Example: Safe Calculator

````python
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    result = num1 / num2
except ValueError:
    print("Please enter valid numbers.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Result:", result)
finally:
    print("Calculator finished.")
````

# Final Summary

Exception handling makes Python programs safer and more reliable.

Remember:

**try** → Try the risky code

**except** → Handle the error

**else** → Run when no error happens

**finally** → Always run
