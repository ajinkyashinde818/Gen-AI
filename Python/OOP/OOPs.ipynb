# Part 1: Types of Constructors

A **constructor** is a special method that runs automatically when an object is created.

In Python, constructors are written using:

```python
__init__()
```


# 1. Default Constructor

A default constructor does not receive additional values while creating an object. It only uses `self`.


````python
class Employee:

    def __init__(self):
        print("Student object created")

student1 = Employee()
````

A default constructor can also assign fixed default values to an object.

````python
class Employee:

    def __init__(self):
        self.name = "Ajinkya"
        self.age = 17

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

student1 = Employee()
student1.display()
````

# 2. Parameterized Constructor

A parameterized constructor receives values when an object is created.

Different objects can receive different values.


````python
class Employee:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

student1 = Employee("Ajinkya", 17)
student2 = Employee("Sakshi", 18)

student1.display()
print()
student2.display()
````

## Constructor Comparison

| Default Constructor | Parameterized Constructor |
|---|---|
| Does not receive additional values | Receives values |
| Uses only `self` | Uses `self` and parameters |
| Can provide fixed/default values | Can provide different values to objects |


# Part 2: Types of Methods

Python OOP has three important types of methods:

1. Instance Method
2. Class Method
3. Static Method


# 1. Instance Method

An instance method works with data belonging to a specific object.

It uses `self` and usually works with instance variables.


````python
class Employee:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

student1 = Employee("Ajinkya", 90)
student1.display()
````

# 2. Class Method

A class method works with the class and usually works with class variables.

It uses `cls` and the `@classmethod` decorator.

`cls` represents the current class.


````python
class Employee:

    college = "MIT AOE"

    @classmethod
    def show_college(cls):
        print("College:", cls.college)

Employee.show_college()
````

# 3. Static Method

A static method does not need object data or class data.

Therefore, it uses neither `self` nor `cls`.

We use the `@staticmethod` decorator.


````python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

result = Calculator.add(10, 20)
print("Result:", result)
````

# self vs cls

```text
self → Current Object
cls  → Current Class
```

- Instance method → uses `self`
- Class method → uses `cls`
- Static method → uses neither


# Example: All Three Methods Together

````python
class Employee:

    college = "MIT AOE"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Instance Method
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

    # Class Method
    @classmethod
    def show_college(cls):
        print("College:", cls.college)

    # Static Method
    @staticmethod
    def welcome():
        print("Welcome to the Student System")

student1 = Employee("Ajinkya", 90)

student1.display()
Employee.show_college()
Employee.welcome()
````

# Difference Between Types of Methods

| Method | Decorator | First Parameter | Mainly Works With |
|---|---|---|---|
| Instance Method | None | `self` | Object / Instance variables |
| Class Method | `@classmethod` | `cls` | Class / Class variables |
| Static Method | `@staticmethod` | None | Independent logic |

## Easy Way to Remember

**Instance Method** → Needs object information

**Class Method** → Needs class information

**Static Method** → Does not need object or class information
