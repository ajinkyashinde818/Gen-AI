# Part 1: OOP Basics

## 1. What is OOP?

OOP stands for **Object-Oriented Programming**.

It is a way of organizing programs using **classes and objects**.

Real-world things have:

- Properties or data
- Actions or behavior

In Python:

- Data is stored using **attributes**
- Actions are written using **methods**


# 2. Class

A **class** is a blueprint or template used to create objects.

For example, one `Student` class can be used to create many student objects.


````python
class Student:
    pass
````

# 3. Object

An **object** is an instance created from a class.

Syntax:

```python
object_name = ClassName()
```

One class can create many objects.


````python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
````

# 4. Attributes

Attributes are variables that store information about an object.

We access them using:

```python
object.attribute
```

````python
class Student:
    name = "Ajinkya"
    age = 19

student1 = Student()

print(student1.name)
print(student1.age)
````

# 5. Methods

A **method** is a function written inside a class.

Methods define what an object can do.

````python
class Student:

    def introduce(self):
        print("Hello, I am a student")

student1 = Student()
student1.introduce()
````

# 6. Understanding self

`self` represents the **current object**.

When you write:

```python
student1.introduce()
```

Python automatically provides the current object to `self`.

Conceptually:

```text
self = student1
```

You normally do not manually pass `self` while calling a method.


# Part 1 Complete Example

````python
class Mobile:

    brand = "Samsung"
    price = 25000

    def show(self):
        print("Brand:", self.brand)
        print("Price:", self.price)

mobile1 = Mobile()
mobile1.show()
````

# Part 2: Constructor and Variables

## 7. Constructor

A constructor is a special method that runs automatically when an object is created.

In Python, the constructor is:

```python
__init__
```

````python
class Student:

    def __init__(self):
        print("Object created")

student1 = Student()
````

# 8. Constructor with Parameters

A constructor can receive data when an object is created.

The received values can be stored inside the object using `self`.


````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Ajinkya", 19)
student2 = Student("Rahul", 20)

print(student1.name)
print(student2.name)
````

## Understanding this line

```python
self.name = name
```

The right side `name` is the value received by the constructor.

The left side `self.name` stores that value inside the current object.

So every object can store different information.


# 9. Instance Variables

An **instance variable** belongs to a specific object.

Example:

```python
self.name
self.marks
```

Different objects can have different values.


````python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Ajinkya", 90)
student2 = Student("Rahul", 75)

print(student1.name, student1.marks)
print(student2.name, student2.marks)
````

# 10. Methods Using Instance Variables

Methods can access instance variables through `self`.


````python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

student1 = Student("Ajinkya", 90)
student1.display()
````

# 11. Class Variables

A **class variable** belongs to the class.

It is generally shared by all objects.

Example:

```python
college = "MIT AOE"
```

````python
class Student:

    college = "MIT AOE"

    def __init__(self, name):
        self.name = name

student1 = Student("Ajinkya")
student2 = Student("Rahul")

print(student1.college)
print(student2.college)
````

# 12. Instance Variable vs Class Variable

| Instance Variable | Class Variable |
|---|---|
| Belongs to an object | Belongs to the class |
| Created using `self` | Created directly inside the class |
| Different objects can have different values | Generally shared by all objects |
| Example: `self.name` | Example: `college` |


# 13. Complete Example

````python
class Car:

    wheels = 4

    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def display(self):
        print("Brand:", self.brand)
        print("Price:", self.price)
        print("Wheels:", self.wheels)

car1 = Car("BMW", 5000000)
car2 = Car("Toyota", 2000000)

car1.display()
print()
car2.display()
````

3. Create a class variable called `bank_name`.

Think carefully about:

- What belongs to each individual object?
- What should be shared by all objects?
- Where should you use `self`?


````python
class Student:
    def __init__(self,name,age,marks):
        self.name=name
        self.age=age
        self.marks=marks
    def display(self):
        print(f"Student Name:",self.name)
        print(f"Student Age:",self.age)
        print(f"Student Marks:",self.marks)

student1=Student("Ajinkya",17,30)
student2=Student("Sakshi",18,30)

student1.display()
print()
student2.display()
````

````python
class BankAccount:
    bank_name="AXIS BANK"

    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance

    def display(self):
        print("Bank Name:",self.bank_name)
        print("Account Holder:",self.account_holder)
        print("Account Balance:",self.balance)

customer1=BankAccount("Tanishka",200000)
customer2=BankAccount("Ajinkya",12345678)

customer1.display()
print()
customer2.display()
````
