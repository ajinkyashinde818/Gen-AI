# Python OOP — Operator Overloading

## What is Operator Overloading?

**Operator overloading** means defining how operators such as `+`, `-`, `*`, `==`, `<`, and `>` should work with objects of our own class.

Python already knows:

```python
10 + 20
```

But if we write:

```python
n1 + n2
```

where `n1` and `n2` are objects of our class, Python needs to know what `+` should mean for those objects.

We define that behavior using **magic methods (dunder methods)**.

## The Basic Idea

The pattern is:

```text
operator
   ↓
magic method
   ↓
our custom behavior
```

For example:

```text
n1 + n2
   ↓
n1.__add__(n2)
```

So `__add__()` tells Python what the `+` operator should do with our objects.

## Example: `__add__()`

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value
```

```python
n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
```

Output:

```text
30
```

### Internally

```text
n1 + n2
   ↓
n1.__add__(n2)
   ↓
self = n1
other = n2
   ↓
self.value + other.value
   ↓
10 + 20
   ↓
30
```

This is operator overloading.

## Why is it called "Overloading"?

The same operator can behave differently depending on the objects.

```python
10 + 20
```

means numerical addition.

```python
"Hello " + "World"
```

means string concatenation.

```python
[1, 2] + [3, 4]
```

means list concatenation.

With our own class, we can define what `+` means by implementing `__add__()`.

> **Operator overloading = giving an operator a custom behavior for objects of our class.**

## Common Operator Overloading Methods

| Operator | Magic Method | Example |
|---|---|---|
| `+` | `__add__()` | `a + b` |
| `-` | `__sub__()` | `a - b` |
| `*` | `__mul__()` | `a * b` |
| `/` | `__truediv__()` | `a / b` |
| `//` | `__floordiv__()` | `a // b` |
| `%` | `__mod__()` | `a % b` |
| `**` | `__pow__()` | `a ** b` |
| `==` | `__eq__()` | `a == b` |
| `!=` | `__ne__()` | `a != b` |
| `<` | `__lt__()` | `a < b` |
| `>` | `__gt__()` | `a > b` |
| `<=` | `__le__()` | `a <= b` |
| `>=` | `__ge__()` | `a >= b` |

## `__sub__()` — Overloading `-`

`__sub__()` controls the `-` operator.

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __sub__(self, other):
        return self.value - other.value
```

```text
n1 - n2
   ↓
n1.__sub__(n2)
```

## `__mul__()` — Overloading `*`

`__mul__()` controls the `*` operator.

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __mul__(self, other):
        return self.value * other.value
```

```text
n1 * n2
   ↓
n1.__mul__(n2)
```

## `__eq__()` — Overloading `==`

`__eq__()` controls the `==` operator.

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks
```

```text
s1 == s2
   ↓
s1.__eq__(s2)
   ↓
compare their marks
```

## Important Pattern

```text
a + b  →  a.__add__(b)
a - b  →  a.__sub__(b)
a * b  →  a.__mul__(b)
a == b →  a.__eq__(b)
a > b  →  a.__gt__(b)
a < b  →  a.__lt__(b)
```

The operator is written normally, and Python calls the corresponding magic method.
