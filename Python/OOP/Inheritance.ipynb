# What is Inheritance?

Inheritance means a **child class can use properties and methods of a parent class**.

Main benefit: **Code Reusability**.

Important syntax:
```python
class Child(Parent):
```

Parent class is also called Base/Super class.
Child class is also called Derived/Sub class.

````python
class Parent:
    def show(self):
        print("This is the Parent class")

class Child(Parent):
    pass

child1 = Child()
child1.show()
````

# Constructors and `super()`

If the child needs its own constructor and also wants to initialize parent data, use:

```python
super().__init__(...)
```

`super()` gives access to the parent class.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)

student1 = Student("Ajinkya", 17, 90)
student1.display()
````

# 1. Single Inheritance

One Parent → One Child

```text
Parent
   ↓
Child
```

````python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

dog1 = Dog()
dog1.eat()
dog1.bark()
````

# 2. Multilevel Inheritance

Inheritance happens through multiple levels.

```text
Grandparent
     ↓
Parent
     ↓
Child
```

````python
class Grandparent:
    def house(self):
        print("Grandparent owns a house")

class Parent(Grandparent):
    def car(self):
        print("Parent owns a car")

class Child(Parent):
    def mobile(self):
        print("Child owns a mobile")

child1 = Child()
child1.house()
child1.car()
child1.mobile()
````

## Practice Question 2: Multilevel Inheritance

Create this structure:

```text
Person
   ↓

```

Requirements:
- `Person` → method `show_person()`
- `Employee` inherits `Person` → method `show_employee()`
- `Manager` inherits `Employee` → method `show_manager()`

# 3. Multiple Inheritance

One child inherits from more than one parent.

```text
Parent1 ──┐
          ↓
        Child
          ↑
Parent2 ──┘
```

Syntax:
```python
class Child(Parent1, Parent2):
```

````python
class Father:
    def father_skill(self):
        print("Father's skill: Programming")

class Mother:
    def mother_skill(self):
        print("Mother's skill: Cooking")

class Child(Father, Mother):
    def child_skill(self):
        print("Child's skill: Gaming")

child1 = Child()
child1.father_skill()
child1.mother_skill()
child1.child_skill()
````

# 4. Hierarchical Inheritance

One parent has multiple children.

```text
        Parent
       /      \\
   Child1    Child2
```

````python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

class Cat(Animal):
    def meow(self):
        print("Cat is meowing")

dog1 = Dog()
cat1 = Cat()

dog1.eat()
dog1.bark()

print()

cat1.eat()
cat1.meow()
````

## Practice Question 4: Hierarchical Inheritance

Create two child classes:

### Circle
- `draw_circle()`

### Rectangle

# 5. Hybrid Inheritance

Hybrid inheritance is a combination of two or more inheritance types.

Example structure:

```text
        A
       / \\
      B   C
       \\ /
        D
```

````python
class A:
    def method_a(self):
        print("Method from A")

class B(A):
    def method_b(self):
        print("Method from B")

class C(A):
    def method_c(self):
        print("Method from C")

class D(B, C):
    def method_d(self):
        print("Method from D")

obj = D()
obj.method_a()
obj.method_b()
obj.method_c()
obj.method_d()
````

## Practice Question 5: Hybrid Inheritance

Create this structure:

```text
        Person
        /    \\
    Student  Employee
        \\    /
         Intern
```

Requirements:

### Person
- `show_person()`

### Student
- Inherits from `Person`
- `study()`

### Employee
- Inherits from `Person`
- `work()`

### Intern
- Inherits from both `Student` and `Employee`
- `intern_work()`

Create an `Intern` object and call all available methods.


````python
class Person:
    def __init__(self):
        print("Hello from person")

    def show_person(self):
        print("Method from Person")

class Student(Person):
    def study(self):
        print("Method from Study")

class Employee(Person):
    def work(self):
        print("Method from work")

class Intern(Student,Employee):
    def intern_work(self):
        print("Method from intern_work")

obj = Intern()
obj.show_person()
obj.study()
obj.work()
obj.intern_work()
````

# Summary

| Type | Structure |
|---|---|
| Single | One Parent → One Child |
| Multilevel | Grandparent → Parent → Child |
| Multiple | Multiple Parents → One Child |
| Hierarchical | One Parent → Multiple Children |
| Hybrid | Combination of inheritance types |

## Important Points

- Parent gives reusable features.
- Child inherits those features.
- `class Child(Parent)` creates inheritance.
- `super().__init__()` calls the parent constructor.
- The main advantage is **code reusability**.
