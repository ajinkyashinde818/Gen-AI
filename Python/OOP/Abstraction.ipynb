- Understand abstraction
- Understand abstract classes and abstract methods
- Use `ABC` and `@abstractmethod`
- Understand why child classes implement abstract methods
- Compare abstraction with encapsulation

> **Abstraction = Hide implementation details and expose essential functionality.**

## 1. What is Abstraction?

Abstraction means hiding unnecessary implementation details and showing only the essential functionality.

### Real-world example: ATM

You can:
- Withdraw money
- Deposit money
- Check balance

You do not need to know the internal processing performed by the ATM.

```text
User → ATM Operation → Internal Processing → Result
```

So, abstraction focuses on **what an object does**, rather than all the details of **how it does it**.

## 2. Abstract Class

An **abstract class** is a class designed mainly as a blueprint/base class for other classes.

Python provides `ABC` through the `abc` module.

```python
from abc import ABC

class Vehicle(ABC):
    pass
```

When an abstract class contains abstract methods, it cannot be instantiated until those methods are implemented by a concrete child class.

## 3. Abstract Method

An **abstract method** is a method declared in an abstract class that child classes are required to implement.

Python uses the `@abstractmethod` decorator.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
```

Here:
- `Vehicle` → abstract class
- `start()` → abstract method
- `@abstractmethod` → marks the method as required
- `pass` → no concrete implementation is provided here

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):

    def start(self):
        print("Car starts with a key")

class Bike(Vehicle):

    def start(self):
        print("Bike starts with a button")

car = Car()
bike = Bike()

car.start()
bike.start()
````

### Output

```text
Car starts with a key
Bike starts with a button
```

The abstract parent defines **what must exist** (`start()`), while each child defines **how it works**.

## 4. Why Can't We Instantiate an Abstract Class?

If `Vehicle` has an unimplemented abstract method:

```python
vehicle = Vehicle()
```

Python raises a `TypeError`.

The class is incomplete because `start()` has not been implemented.

The concrete child classes provide the implementation.

## 5. Multiple Abstract Methods

An abstract class can contain multiple abstract methods.

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

    @abstractmethod
    def move(self):
        pass
```

A concrete child class must implement all required abstract methods before its object can be created.

## 6. Abstract Class vs Abstract Method

| Concept | Meaning |
|---|---|
| Abstract Class | Blueprint/base class for child classes |
| Abstract Method | Method that concrete child classes must implement |
| `ABC` | Base class used for abstract classes |
| `@abstractmethod` | Decorator used to declare an abstract method |

### Exam definitions

**Abstract Class:** A class that serves as a blueprint for derived classes and cannot be instantiated while it has unimplemented abstract methods.

**Abstract Method:** A method declared using `@abstractmethod` that must be implemented by concrete child classes.

## 7. Abstraction vs Encapsulation

| Encapsulation | Abstraction |
|---|---|
| Controls access to data | Hides implementation complexity |
| Protects/controls data | Exposes essential functionality |
| Private attributes, getters, setters | Abstract classes, abstract methods |
| Example: `__balance` | Example: `Vehicle.start()` |

### Memory trick

```text
ENCAPSULATION → Protect / control data
ABSTRACTION   → Hide complexity
```
