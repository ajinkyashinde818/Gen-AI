## What is Polymorphism?
**Polymorphism** means **"many forms."**

The same method, function, or interface can behave differently depending on the object or situation.

> **Same method → Different objects → Different behavior**

## 1. Method Overriding

Method overriding occurs when a child class provides its own implementation of a method already defined in the parent class.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")
```

The `Dog` version replaces the inherited behavior when `sound()` is called on a `Dog` object.

````python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

animal = Animal()
dog = Dog()

animal.sound()
dog.sound()
````

## 2. Polymorphism with Multiple Child Classes

Different child classes can override the same parent method differently.

````python
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

class Cat(Animal):
    def sound(self):
        print("Meow")

animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
````

## 3. `super()` with Method Overriding

`super()` lets a child method call the parent implementation and then add its own behavior.

````python
class Parent:
    def show(self):
        print("Parent method")

class Child(Parent):
    def show(self):
        super().show()
        print("Child method")

obj = Child()
obj.show()
````

## 4. Duck Typing

Python often follows **duck typing**:

> If an object provides the required behavior, we can use it.

The classes do not need to inherit from the same parent.

````python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

def make_sound(animal):
    animal.sound()

make_sound(Dog())
make_sound(Cat())
````

## 5. Operator Polymorphism

The same operator can behave differently for different data types.

- `10 + 20` → `30`
- `"10" + "20"` → `"1020"`
- `3 * 4` → `12`
- `"Hi " * 3` → `"Hi Hi Hi "`

````python
print(10 + 20)
print("10" + "20")
print(3 * 4)
print("Hi " * 3)
````

## 6. Function Polymorphism

Built-in functions can work with different types.

For example, `len()` works with strings, lists, tuples, and other objects that support the required behavior.

````python
print(len("Python"))
print(len([10, 20, 30, 40]))
print(len((1, 2, 3)))
````

## 7. Method Overloading in Python

Python does **not** support traditional method overloading by defining the same method multiple times with different parameter lists. The later definition replaces the earlier one.

Instead, use techniques such as default arguments or `*args`.

Example with a default argument:

````python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
````

## 8. Complete Polymorphism Example

A major advantage of polymorphism is that code can use a common method without checking the exact object type.

```python
animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
```

The loop simply calls `sound()`. Each object decides how that method behaves.

# Quick Revision

| Concept | Meaning |
|---|---|
| Polymorphism | One interface/method with many forms |
| Method overriding | Child changes a parent method's implementation |
| `super()` | Calls the parent implementation |
| Duck typing | Focuses on what an object can do |
| Operator polymorphism | Same operator, different behavior |
| Function polymorphism | Same function works with different object types |

## Problem 1 — Payment System

Create a parent class `Payment` with a method `pay()`.

Create three child classes:
- `UPIPayment`
- `CardPayment`
- `CashPayment`

Override `pay()` in each class with a different message.

Then:
1. Create one object of each class.
2. Store them in a list.
3. Use a loop to call `pay()` for every object.

**Concept:** Method overriding + polymorphism.

````python
class Payment:
    def pay(self):
        print("This is Parent Class")

class UPIPayment(Payment):
    def pay(self):
        print("This is UPI Payment class")

class CardPayment(Payment):
    def pay(self):
        print("This is Card Payment Class")

class CashPayment(Payment):
    def pay(self):
        print("This is a Cash Payment Class")

payment=[UPIPayment(),CardPayment(),CashPayment()]

for pmethod in payment:
    pmethod.pay()
````

## Problem 2 — Shape Area

Create a parent class `Shape` with a method `area()`.

Create:
- `Rectangle`
- `Circle`

Requirements:
- Rectangle calculates area using length and width.
- Circle calculates area using radius.
- Override `area()` in both child classes.
- Store both objects in a list.
- Use a loop to call `area()`.

**Concept:** Polymorphism with inheritance.

**Hint:** The loop should not need to know whether the object is a rectangle or circle.

````python
class Shape:
    def area(self):
        print("This is Parent Class")

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        print("The Area of Rectangle:", self.length * self.width)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print("The Area of Circle:", 3.14 * self.radius * self.radius)

Areas=[Rectangle(5,4),Circle(5)]

for areas in Areas:
    areas.area()
````

## Problem 3 — Duck Typing

Create three unrelated classes:
- `Dog`
- `Cat`
- `Cow`

Each class should have a `sound()` method.

Create:

```python
def make_sound(animal):
    ...
```

The function should call the object's `sound()` method.

Pass objects of all three classes to the same function.

**Important:** Do not use inheritance between these classes.

**Concept:** Duck typing.

````python
class Dog:
    def sound(self):
        print("Dog says: Woof")


class Cat:
    def sound(self):
        print("Cat says: Meow")


class Cow:
    def sound(self):
        print("Cow says: Moo")


def make_sound(animal):
    animal.sound()


dog = Dog()
cat = Cat()
cow = Cow()

make_sound(dog)
make_sound(cat)
make_sound(cow)
````
