> **Magic methods** are special methods whose names start and end with double underscores, such as `__init__()` and `__str__()`.

## 1. What Are Magic Methods?

Magic methods are special methods that allow Python objects to work with built-in operations.

They are also called **dunder methods** because of the double underscores.

Examples:

```python
__init__()
__str__()
__len__()
__add__()
__eq__()
```

Python often calls them automatically.

For example:

```python
print(obj)
```

can use:

```python
obj.__str__()
```

And:

```python
a + b
```

can use:

```python
a.__add__(b)
```

## 2. `__init__()` — Object Initialization

`__init__()` is automatically called when an object is created.

```python
class Student:

    def __init__(self, name):
        self.name = name
```

When:

```python
s1 = Student("Rahul")
```

Python automatically calls `__init__()`.

```text
Student("Rahul")
       ↓
   __init__()
       ↓
self.name = "Rahul"
```

## 3. `__str__()` — String Representation

`__str__()` defines a readable string representation of an object.

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"Name: {self.name}, Marks: {self.marks}"
```

Then:

```python
s1 = Student("Rahul", 90)
print(s1)
```

Output:

```text
Name: Rahul, Marks: 90
```

**Important:** `__str__()` should return a string.

## 4. `__len__()` — `len()` Function

`__len__()` defines what happens when `len()` is used on your object.

```python
class Team:

    def __init__(self, players):
        self.players = players

    def __len__(self):
        return len(self.players)
```

Then:

```python
team = Team(["A", "B", "C", "D"])
print(len(team))
```

Output:

```text
4
```

Conceptually:

```text
len(team)
    ↓
team.__len__()
    ↓
4
```

## 5. `__add__()` — `+` Operator

`__add__()` allows you to define how two objects behave with `+`.

```python
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value
```

Then:

```python
n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
```

Output:

```text
30
```

Conceptually:

```text
n1 + n2
   ↓
n1.__add__(n2)
   ↓
30
```

This is called **operator overloading**.

## 6. `__eq__()` — Equality Operator

`__eq__()` controls how two objects behave with `==`.

```python
class Student:

    def __init__(self, roll):
        self.roll = roll

    def __eq__(self, other):
        return self.roll == other.roll
```

Then:

```python
s1 = Student(10)
s2 = Student(10)

print(s1 == s2)
```

Output:

```text
True
```

Conceptually:

```text
s1 == s2
   ↓
s1.__eq__(s2)
   ↓
True
```

## 7. Common Magic Methods

| Magic Method | Operation |
|---|---|
| `__init__()` | Object initialization |
| `__str__()` | `print(object)` / readable string |
| `__repr__()` | Developer-oriented representation |
| `__len__()` | `len(object)` |
| `__add__()` | `+` |
| `__sub__()` | `-` |
| `__mul__()` | `*` |
| `__truediv__()` | `/` |
| `__eq__()` | `==` |
| `__ne__()` | `!=` |
| `__lt__()` | `<` |
| `__gt__()` | `>` |
| `__le__()` | `<=` |
| `__ge__()` | `>=` |

## 8. Magic Methods and Operator Overloading

Magic methods let us customize operators for our own classes.

```text
+    → __add__()
-    → __sub__()
*    → __mul__()
/    → __truediv__()
==   → __eq__()
<    → __lt__()
>    → __gt__()
<=   → __le__()
>=   → __ge__()
```

For example:

```python
obj1 + obj2
```

can invoke:

```python
obj1.__add__(obj2)
```

This makes custom objects behave more naturally.

# Practice Task 3 — Compare Students

Create a `Student` class.

### Requirements
1. Create `__init__(self, name, marks)`.
2. Store `name` and `marks`.
3. Implement `__eq__(self, other)`.
4. Two students should be equal when their marks are equal.
5. Create two students with marks `85`.
6. Check `student1 == student2`.
7. Change the second student's marks and test again.

**Goal:** Practice `__eq__()` and object comparison.

Build it yourself and send me your code for review.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks


student1 = Student("Ajinkya", 85)
student2 = Student("Aarya", 85)

print("Are students equal?", student1 == student2)

student2.marks = 90

print("After changing second student's marks:")
print("Are students equal?", student1 == student2)
````
