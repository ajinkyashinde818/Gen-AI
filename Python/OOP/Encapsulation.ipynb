## 1. What is Encapsulation?

**Encapsulation** means bundling data and the methods that operate on that data inside a class, while controlling how the data is accessed or modified.

Main idea:

> **Protect internal data and provide controlled access to it.**

For example, a bank account should not allow its balance to be changed to an invalid value directly.

Instead, the class can provide:
- `get_balance()` → read the balance
- `set_balance()` → change the balance

## 2. Private Attributes

In Python, a double underscore (`__`) is commonly used to indicate a private attribute.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
```

The attribute is intended to be accessed through methods provided by the class rather than directly from outside.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

account = BankAccount(5000)
print("Bank account created.")
````

## 3. Getter Method

A **getter** is a method used to read or retrieve the value of a private attribute.

```python
def get_balance(self):
    return self.__balance
```

Usage:

```python
account.get_balance()
```

Think:

`get_balance()` → **READ** the private data.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)
print(account.get_balance())
````

## 4. Setter Method

A **setter** is a method used to change the value of a private attribute.

```python
def set_balance(self, balance):
    self.__balance = balance
```

Usage:

```python
account.set_balance(8000)
```

Think:

`set_balance()` → **MODIFY** the private data.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):
        self.__balance = balance

account = BankAccount(5000)

print(account.get_balance())
account.set_balance(8000)
print(account.get_balance())
````

## 5. Setter with Validation

The real benefit of a setter is that it can validate data before changing the private attribute.

For example, a bank balance should not be negative.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
        else:
            print("Balance cannot be negative")

account = BankAccount(5000)

account.set_balance(8000)
print(account.get_balance())

account.set_balance(-1000)
print(account.get_balance())
````

## 6. Real-World Example — Student Marks

Marks should normally be between `0` and `100`.

A setter can prevent invalid marks from being stored.

````python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")

student = Student(85)

print(student.get_marks())

student.set_marks(95)
print(student.get_marks())

student.set_marks(150)
print(student.get_marks())
````

## 7. Getter vs Setter

| Method | Purpose | Example |
|---|---|---|
| Getter | Read data | `get_marks()` |
| Setter | Modify data | `set_marks(90)` |

### Key idea

Without controlled access:

```python
object.data = value
```

With a getter/setter approach:

```python
object.get_data()
object.set_data(value)
```

A setter can validate, restrict, or process a new value before storing it.
