# Map function()

## What is map()?

`map()` applies the same function to every element of an iterable.

**Syntax**

```python
map(function, iterable)
```

It returns a map object, so use `list()` to display the values.


````python
a=[1,2,3,4]
def square(x):
    return x**2

value=list(map(square,a))
print(value)
````

````python
a=list(map(lambda x:x*2,[1,2,3,4]))

print(a)
````

````python
# Map function with multiple itterables

l1=[1,2,3,4]
l2=[2,4,6,8]

print(list(map(lambda a,b:a+b,l1,l2)))
````

## Practice Questions

1. Multiply every number by 5.

3. Convert names to uppercase.
4. Add 10 to every number.
5. Convert Celsius to Fahrenheit using map().
6. Convert string numbers into integers.

````python
# 1.

a=[1,2,3,4,5]

def multiply(x):
    return x*5

value=list(map(multiply,a))
print("Multiply by 5",value)
````

````python
# 2.

a=[1,2,3,4]
def square(x):
    return x**2

value=list(map(square,a))
print(value)
````

````python
# 3.
name=['Ajinkya','Nana','Deva','Nandu','Taru']
def convert(x):
    return x.lower()

value=list(map(convert,name))
print(value)
````

````python
# 4.

print(list(map(lambda x:x+10,[1,20,3,4,5])))
````

````python
# 5.

print(list(map(lambda x:(x*1.8)+32,[15])))
````

````python
# 6.

num=['1','2','3','4','5']

def conversion(x):
    return int(x)

print(list(map(conversion,num)))
````
