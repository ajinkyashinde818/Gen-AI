# Filter Function

### A Filter function is going to retrive a only this item which are meeting a certain condition in an iterable.

### Syntax:

    list(filter(function,iterable))

````python
num=[1,2,3,4,5,6,7,8,9,0]

def even(value):
    if value%2==0:
        return True

even_or_not=list(filter(even,num))
print("Even Numbers:",even_or_not)
````

````python
num=[1,2,3,4,5,6,7,8,9,0]

def even(value):
    if value%2!=0:
        return True

even_or_not=list(filter(even,num))
print("Even Numbers:",even_or_not)
````

````python
marks=[20,45,67,12,90,35]

def pass_marks(mark):
    return mark>=35

print(list(filter(pass_marks,marks)))
````

## Difference Between map() and filter()

| map() | filter() |
|-------|----------|
| Changes every element | Selects matching elements |
| Function returns new value | Function returns True/False |
| Output size usually same | Output may be smaller |
