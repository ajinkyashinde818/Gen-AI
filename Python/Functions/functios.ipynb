# Python Functions
A complete guide with examples.

## Syntax
```python
def function_name(parameters):
    # code
```
Functions can contain normal Python code including input(), loops, if-else, lists, etc.

````python
def greet():
    print("Hello, Welcome to Python!")

greet()
````

## User Input Inside Function

````python
def add():
    a=int(input("Enter first number: "))
    b=int(input("Enter second number: "))
    print("Sum =",a+b)

add()
````

## Parameters and Arguments

````python
def add(a,b):
    return a+b

x=int(input("Enter first number: "))
y=int(input("Enter second number: "))
print("Sum =",add(x,y))
````

## Return Statement

````python
def square(n):
    return n*n

print(square(5))
````

## Default Parameter

````python
def greet(name="Guest"):
    print("Hello",name)

greet()
greet("Ajinkya")
````

## Local and Global Variables

````python
x=100

def demo():
    y=10
    print("Global:",x)
    print("Local:",y)

demo()
print(x)
````

## Four Types of Functions

````python
# No arguments, No return
def hello():
    print("Hello")
hello()

# Arguments, No return
def show(name):
    print(name)
show("Ajinkya")

# No arguments, Return
def value():
    return 50
print(value())

# Arguments, Return
def multiply(a,b):
    return a*b
print(multiply(5,6))
````
