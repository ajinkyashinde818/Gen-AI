## Four Types of Functions

````python
# No arguments, No return
def hello():
    print("Hello")
hello()

# Arguments, No return
def show(name):
    print(name)
show("Ajinkya")

# No arguments, Return
def value():
    return 50
print(value())

# Arguments, Return
def multiply(a,b):
    return a*b
print(multiply(5,6))
````

````python
# 8.

def reverse(num):
    rev = 0

    while num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num = num // 10

    print("Reversed Number =", rev)

number = int(input("Enter a Number: "))
reverse(number)
````

````python
# 9.

def count_vowels(text):
    count = 0

    for ch in text.lower():
        if ch in "aeiou":
            count += 1

    print("Total Vowels =", count)

name = input("Enter a String: ")
count_vowels(name)
````

````python
'''Problem Statement: Student Grade Calculator

Write a Python program using functions to calculate a student's percentage and display the grade.

Requirements:
Create a function named percentage(total_marks, obtained_marks) that returns the percentage.
Create another function named grade(percentage) that returns:
A if percentage is 90% or above
B if percentage is 75% to 89%
C if percentage is 50% to 74%
Fail if percentage is below 50%
Take the total marks and obtained marks from the user.
Display:
Percentage
Grade
Sample Output
Enter Total Marks: 500
Enter Obtained Marks: 445

Percentage = 89.0%
Grade = B'''
````

````python
def percentage(total_marks, obtained_marks):
    print("Total Marks:",total_marks)
    print("Obtained Marks:",obtained_marks)
    
    return (obtained_marks/total_marks)*100
    
def grade(percentage):
    if percentage>=90:
        return "A"
    elif percentage<=89 and percentage>=75:
        return "B"
    elif percentage<=74 and percentage>=50:
        return "C"
    else:
        return "Fail"
    

per=percentage(500,445)
gra=grade(per)
    
print("Percentage =", per)
print("Grade =", gra)
````
