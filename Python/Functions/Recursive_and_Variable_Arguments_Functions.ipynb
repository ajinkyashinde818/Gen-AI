## Recursive Functions
A recursive function calls itself until a base condition is met.

```python
def print_num(num):
    if num>5:
        return
    print(num)
    print_num(num+1)

print_num(1)
```

```python
def factorial(n):
    if n==1:
        return 1
    return n*factorial(n-1)

print(factorial(5))
```

## Variable Arguments (`*args`)

```python
def add(*numbers):
    total=0
    for i in numbers:
        total+=i
    print(total)

add(10,20)
add(10,20,30,40)
```

## Keyword Variable Arguments (`**kwargs`)

```python
def student(**details):
    for k,v in details.items():
        print(k,":",v)

student(name="Ajinkya",age=20,city="Pune")
```
