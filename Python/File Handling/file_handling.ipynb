# File Handling

## A File handling means creating ,reading,writing and modify the file using the python.

### Think of a file as a notebook.

    ✍️ Write → Add notes.
    📖 Read → Read notes.
    ➕ Append → Add more notes without removing old ones.
    ❌ Delete → Remove the notebook.

# File Handling Steps

1. Open File
2. Read / Write / Append
3. Close File

**Syntax**

```python
open(filename, mode)
```


| Mode   | Meaning         |
| ------ | --------------- |
| `"r"`  | Read            |
| `"w"`  | Write           |
| `"a"`  | Append          |
| `"x"`  | Create new file |
| `"r+"` | Read + Write    |
| `"w+"` | Write + Read    |
| `"a+"` | Append + Read   |


````python
file=open("student.txt","r")
print(file.read())
file.close()
````

````python
file=open("student.txt","w")
file.write("I am Ajinkya")
file.close()
````

````python
file=open("student.txt","a")
file.write("Hello")
file.close()
````

# Detailed File Modes


| Mode | Description | File Exists? | Example |
|------|-------------|--------------|---------|
| `r` | Read only | Required | `open("data.txt","r")` |
| `w` | Write (overwrites file) | Optional | `open("data.txt","w")` |
| `a` | Append data | Optional | `open("data.txt","a")` |
| `x` | Create new file | Must not exist | `open("data.txt","x")` |
| `r+` | Read and Write | Required | `open("data.txt","r+")` |
| `w+` | Write and Read | Optional | `open("data.txt","w+")` |
| `a+` | Append and Read | Optional | `open("data.txt","a+")` |


## Example of `r` Mode

````python
file=open("student.txt","r")
print(file.read())
file.close()
````

## Example of `w` Mode

````python
file=open("student.txt","w")
file.write("Python File Handling")
file.close()
````

## Example of `a` Mode

````python
file=open("student.txt","a")
file.write("\nNew Line")
file.close()
````

## Example of `x` Mode

````python
# Run only if file does not exist
file=open("newfile.txt","x")
file.close()
````

## Example of `r+` Mode

````python
file=open("student.txt","r+")
print(file.read())
file.write("\nEnd")
file.close()
````

## Example of `w+` Mode

````python
file=open("student.txt","w+")
file.write("Hello")
file.seek(0)
print(file.read())
file.close()
````

## Example of `a+` Mode

````python
file=open("student.txt","a+")
file.write("\nPython")
file.seek(0)
print(file.read())
file.close()
````
