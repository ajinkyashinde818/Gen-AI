# OS Library

### The OS (Operating System) module is a built-in Python library that provides functions to interact with the operating system.

## Commonly Used Functions

| Function        | Purpose               |
| --------------- | --------------------- |
| `getcwd()`      | Current directory     |
| `chdir()`       | Change directory      |
| `listdir()`     | List files/folders    |
| `mkdir()`       | Create folder         |
| `makedirs()`    | Create nested folders |
| `rmdir()`       | Remove empty folder   |
| `removedirs()`  | Remove nested folders |
| `remove()`      | Delete file           |
| `rename()`      | Rename file/folder    |
| `path.exists()` | Check existence       |
| `path.isfile()` | Check file            |
| `path.isdir()`  | Check folder          |
| `path.join()`   | Join paths            |
| `walk()`        | Traverse folders      |
| `environ`       | Environment variables |
| `system()`      | Execute OS command    |

# Advanced `os` Module Functions

## `os.scandir()`
Explanation and example:

```python
import os
with os.scandir('.') as entries:
    for e in entries:
        print(e.name, e.is_file(), e.is_dir())
```

## `os.stat()`
Explanation and example:

```python
import os
info=os.stat('student.txt')
print(info)
```
