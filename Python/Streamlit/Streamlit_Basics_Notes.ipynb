## 1. What is Streamlit?

**Streamlit is a Python library used to create web applications using Python.**

The important point is:

> You can build an interactive web application without writing HTML, CSS, or JavaScript.

A traditional web application may involve:

```text
HTML
CSS
JavaScript
Backend
API
```

With Streamlit, much of the interface can be created directly with Python.


## 2. Why is Streamlit useful for Data Science?

Streamlit is especially useful for **Data Science and AI/ML projects**.

A typical machine-learning workflow is:

```text
Dataset
   ↓
Data Cleaning
   ↓
Feature Engineering
   ↓
Machine Learning Model
   ↓
Prediction
```

Streamlit can provide a user interface for the model:

```text
        AI Prediction App

Enter Age:       [ 21 ]

Enter Salary:    [ 50000 ]

Experience:      [ 2 ]

             [ Predict ]

Prediction: Eligible
```

So:

**Python + Machine Learning + Streamlit = Interactive AI/ML Application**


## 3. Installation

Open CMD, PowerShell, or a terminal and run:

```bash
pip install streamlit
```

Check the installation:

```bash
streamlit --version
```

> Note: These commands are normally run in a terminal, not as regular Python code.


## 4. Your First Streamlit Application

Create a Python file named:

```text
app.py
```

Put this code inside it:

```python
import streamlit as st

st.title("My First Streamlit App")
st.write("Hello World!")
```

Run it from the terminal:

```bash
streamlit run app.py
```

Streamlit will start a local web server and normally open the application in your browser.

You will usually see a URL similar to:

```text
http://localhost:8501
```


## 5. Understanding `import streamlit as st`

```python
import streamlit as st
```

Here:

```text
streamlit
    ↓
Python library

st
    ↓
short name / alias
```

Instead of writing:

```python
streamlit.title()
streamlit.write()
streamlit.button()
```

we normally write:

```python
st.title()
st.write()
st.button()
```


## 6. `st.title()`

`st.title()` is used to display the main title of the application.

Example:

```python
import streamlit as st

st.title("Student Management System")
```

Think of it as:

```text
st.title()
    ↓
Main application title
```


## 7. `st.header()`

`st.header()` is used for section headings.

Example:

```python
st.header("Student Information")
```

Complete example:

```python
import streamlit as st

st.title("Student Management System")
st.header("Student Information")
```


## 8. `st.subheader()`

`st.subheader()` is used for smaller section headings.

Example:

```python
st.subheader("Personal Details")
```

A simple hierarchy is:

```text
title
  ↓
header
  ↓
subheader
```


## 9. `st.write()`

`st.write()` is one of the most useful Streamlit functions.

It can display different types of content.

### Text

```python
st.write("Welcome to my application")
```

### Number

```python
st.write(100)
```

### Variable

```python
name = "Ajinkya"
st.write(name)
```

### Calculation

```python
a = 10
b = 20

st.write(a + b)
```

Output:

```text
30
```


## 10. Markdown

Streamlit supports Markdown formatting.

Example:

```python
st.markdown("## Welcome")
```

Common Markdown syntax:

```text
# Heading 1
## Heading 2
### Heading 3

**Bold**

*Italic*
```

Example:

```python
st.markdown("# Student Dashboard")
st.markdown("**Computer Science & Engineering**")
```


## 11. Complete Basic Example

```python
import streamlit as st

st.title("Student Management System")

st.header("Student Information")

st.subheader("Personal Details")

name = "Ajinkya"
branch = "CSE (Data Science)"
year = "Second Year"

st.write("Name:", name)
st.write("Branch:", branch)
st.write("Year:", year)

st.markdown("**Welcome to my Streamlit application!**")
```

Run the file with:

```bash
streamlit run app.py
```


## 12. How Streamlit Works

A normal Python program usually works like:

```text
Python file
     ↓
Execute
     ↓
Terminal output
```

A Streamlit application works like:

```text
Python file
     ↓
Streamlit
     ↓
Web application
     ↓
Browser
```

This is why Streamlit is very useful for **Data Science, Machine Learning, and AI projects**.
