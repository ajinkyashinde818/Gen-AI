# Python Dictionary — Detailed Notes and Examples

This notebook covers dictionary concepts from basic to advanced with runnable examples.

## What is a Dictionary?
A dictionary stores data in **key-value pairs**.

- Syntax: `{key: value}`
- Mutable: Yes
- Keys: Unique
- Values: Can repeat
- Access: Usually by key


````python
student = {
    "name": "Ajinkya",
    "age": 21,
    "branch": "Computer Technology",
    "marks": 90
}

print(student)
````

## 1. Key and Value

In the dictionary below:
- `"name"` is a key and `"Ajinkya"` is its value.
- `"age"` is a key and `21` is its value.


````python
student = {
    "name": "Ajinkya",
    "age": 21,
    "marks": 90
}

print(student)
````

## 2. Accessing Values Using Keys


````python
print(student["name"])
print(student["marks"])
````

## 3. Dictionary is Mutable

A dictionary can be changed after creation.


````python
student["marks"] = 95
print(student)
````

## 4. Adding a New Key-Value Pair


````python
student["branch"] = "Computer Technology"
print(student)
````

## 5. Change vs Add

- Existing key → update its value
- New key → add a new pair


````python
student["marks"] = 98       # Update
student["city"] = "Sangamner"  # Add
print(student)
````

## 6. `get()` Method


````python
print(student.get("name"))
print(student.get("phone"))
print(student.get("phone", "Not Available"))
````

## 7. `keys()` Method


````python
print(student.keys())
````

## 8. `values()` Method


````python
print(student.values())
````

## 9. `items()` Method


````python
print(student.items())
````

## 10. Loop Through Dictionary Keys


````python
for x in student:
    print(x)
````

## 11. Loop Through Values


````python
for x in student.values():
    print(x)
````

## 12. Loop Through Keys and Values Together


````python
for key, value in student.items():
    print(key, value)
````

## 13. Check Whether a Key Exists


````python
if "name" in student:
    print("Name exists")

print("Ajinkya" in student)
print("Ajinkya" in student.values())
````

## 14. Remove Item Using `pop()`


````python
data = {
    "name": "Ajinkya",
    "age": 21,
    "marks": 90
}

data.pop("age")
print(data)
````

## 15. Remove Last Inserted Pair Using `popitem()`


````python
data = {
    "name": "Ajinkya",
    "age": 21,
    "marks": 90
}

data.popitem()
print(data)
````

## 16. Delete an Item Using `del`


````python
data = {
    "name": "Ajinkya",
    "age": 21,
    "marks": 90
}

del data["age"]
print(data)
````

## 17. Remove All Items Using `clear()`


````python
data = {
    "name": "Ajinkya",
    "age": 21
}

data.clear()
print(data)
````

## 18. `update()` Method


````python
student = {
    "name": "Ajinkya",
    "age": 21
}

student.update({
    "age": 22,
    "marks": 90
})

print(student)
````

## 19. Duplicate Keys Are Not Allowed

If a key appears again, the later value replaces the earlier value.


````python
student = {
    "name": "Ajinkya",
    "marks": 80,
    "marks": 95
}

print(student)
````

## 20. Duplicate Values Are Allowed


````python
students = {
    "Ajinkya": 90,
    "Rahul": 90,
    "Sakshi": 90
}

print(students)
````

## 21. Different Data Types as Values


````python
student = {
    "name": "Ajinkya",
    "age": 21,
    "marks": 90.5,
    "passed": True,
    "subjects": ["Python", "Java"]
}

print(student)
````

## 22. Rules for Dictionary Keys

Common valid keys include strings, integers, and suitable tuples.
A list cannot be used as a dictionary key.


````python
data = {
    "name": "Ajinkya",
    1: "One",
    (10, 20): "Point"
}

print(data)
````

## 23. Nested Dictionary


````python
students = {
    "student1": {
        "name": "Ajinkya",
        "marks": 90
    },
    "student2": {
        "name": "Rahul",
        "marks": 80
    }
}

print(students["student1"]["name"])
print(students["student2"]["marks"])
````

## 24. List of Dictionaries


````python
students = [
    {"name": "Ajinkya", "marks": 90},
    {"name": "Rahul", "marks": 80},
    {"name": "Sakshi", "marks": 95}
]

print(students[1]["name"])
````

## 25. Dictionary Comprehension


````python
squares = {x: x * x for x in range(1, 6)}
print(squares)
````

### Dictionary Comprehension Pattern

```python
{key_expression: value_expression for item in iterable}
```


## 26. Copying a Dictionary


````python
student = {
    "name": "Ajinkya",
    "marks": 90
}

new_student = student.copy()
print(new_student)
````

## 27. Dictionary Length


````python
student = {
    "name": "Ajinkya",
    "age": 21,
    "marks": 90
}

print(len(student))
````

## 28. List vs Tuple vs Dictionary

| Feature | List | Tuple | Dictionary |
|---|---|---|---|
| Syntax | `[]` | `()` | `{key: value}` |
| Data style | Values | Values | Key-value pairs |
| Mutable | Yes | No | Yes |
| Access | Index | Index | Key |
| Duplicates | Yes | Yes | Keys: No, Values: Yes |

## Final Summary

A dictionary is a **mutable data structure** that stores information in **key-value pairs**.

Important concepts:
1. Create a dictionary
2. Access values using keys
3. Add and update items
4. Use `get()`
5. Use `keys()`, `values()`, and `items()`
6. Loop through dictionaries
7. Remove items
8. Nested dictionaries
9. List of dictionaries
10. Dictionary comprehension
