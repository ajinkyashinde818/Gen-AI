# Tuples

## Tuple are the type of Data structure that stored a element in a ordered way just like a list nut it can be immutable. means once create not change or add.

```python
car=("Swift","Fortuner","Ferreri","GT 650")
print(car)
print(type(car))
```

```python
student = "Ajinkya", 21, 85.5

print(student)
print(type(student))
```

## Tuples stored a data in ordered manner and support any type of data

```python
num=(2,4,3,1,5)
print(num)

data=("Ajinkya",45.65,False,43)
print(data)
```

```python
fruits = ("Apple", "Banana", "Mango", "Guava", "Grapes")

print(fruits[1:4])
print(fruits[1:4:2])
```

```python
fruits = ("Apple", "Banana", "Mango")

fruits[1]="orange"

print(fruits)
#  'tuple' object does not support item assignment
```

```python
bike=("GT 650")
print(type(bike))

bike=("GT 650",)
print(type(bike))
```

## Tuple Unpacking

Tuple unpacking means taking tuple values and storing them in separate variables.

```python
perform=("Ajinkya",47,91.89)

name,rollno,percent=perform
print(name)
```

```python
fruits = ("Apple", "Banana", "Mango")

for fruit in fruits:
    print(fruit)
```

## Tuples has 2 methods 

1.count
2.index

```python
numbers = (10, 20, 10, 30, 10)
print(numbers.count(10))

fruits=("Apple","Banana","Orange")
print(fruits.index("Banana"))
```

## Packing and unpacking of Tuples

```python
# Packing

abc=("Harrry","Bhabad","Supriya")
print(abc)

#unpacking
a,b,c,d,e=(1,2,3,4,5)
print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
# unpacking with *

a,*b,c=(1,2,3,4,5,6,7,8,9)
print(a)
print(b)
print(c)
```

```python
a=(1,2,3)
b=(4,5,6)
print(a+b)
```

```python
# Converting List to Tuple

lists=[1,2,3,4,5]
conversion=tuple(lists)
print(conversion)
```

```python
# Converting Tuple to List

tuples=(1,2,3,4,5)
conversion=list(tuples)
print(conversion)
```

```python
students = (
    ("Ajinkya", 90),
    ("Rahul", 80),
    ("Priya", 95)
)

print(students)
```

## List vs Tuple
| Feature        | List          | Tuple      |
| -------------- | ------------- | ---------- |
| Syntax         | `[]`          | `()`       |
| Mutable        | ✅ Yes         | ❌ No       |
| Ordered        | ✅ Yes         | ✅ Yes      |
| Indexed        | ✅ Yes         | ✅ Yes      |
| Duplicates     | ✅ Yes         | ✅ Yes      |
| Slicing        | ✅ Yes         | ✅ Yes      |
| Usually faster | Less          | More       |
| Methods        | Many          | Few        |
| Best for       | Changing data | Fixed data |
