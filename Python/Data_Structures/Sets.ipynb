# Python Sets — Detailed Notes

This notebook explains **Python Sets** from basic to advanced with simple examples.

A **set** is a collection of **unique values**.


## 1. Basic Set Example

```python
numbers = {10, 20, 30, 40}
print(numbers)
```

## 2. Main Properties of Set

A Python set is:

- **Unordered** → no fixed position
- **Unique** → duplicate values are removed
- **Mutable** → we can add/remove values
- **Not indexed** → `set[0]` is invalid


```python
numbers = {10, 20, 20, 30, 30, 30}
print(numbers)
```

## 3. Creating a Set

```python
fruits = {"apple", "banana", "mango"}
print(fruits)

data = {10, 20.5, "Ajinkya", True}
print(data)
```

## 4. Empty Set — Important

`{}` creates an empty dictionary, not an empty set.

Use `set()` for an empty set.


```python
data = set()
print(data)
print(type(data))
```

## 5. Set Does Not Allow Duplicates

```python
marks = {80, 90, 80, 95, 90}
print(marks)

students = ["Ajinkya", "Rahul", "Ajinkya", "Sakshi", "Rahul"]
unique_students = set(students)
print(unique_students)
```

## 6. Accessing Set Elements

Sets do not support indexing because they have no fixed index.

Use a loop to access elements.


```python
fruits = {"apple", "banana", "mango"}

for fruit in fruits:
    print(fruit)
```

## 7. Add One Element — `add()`

```python
fruits = {"apple", "banana"}
fruits.add("mango")
print(fruits)

numbers = {10, 20}
numbers.add(30)
print(numbers)
```

## 8. Add Multiple Elements — `update()`

```python
numbers = {10, 20}
numbers.update([30, 40, 50])
print(numbers)

a = {1, 2, 3}
b = {4, 5, 6}
a.update(b)
print(a)
```

## 9. Remove Element — `remove()`

`remove()` deletes an element.

If the value does not exist, Python gives an error.


```python
numbers = {10, 20, 30}
numbers.remove(20)
print(numbers)
```

## 10. Remove Safely — `discard()`

`discard()` does not give an error if the value is missing.


```python
numbers = {10, 20, 30}
numbers.discard(100)
print(numbers)
```

### `remove()` vs `discard()`

| Method | Missing value |
|---|---|
| `remove()` | Error |
| `discard()` | No error |


## 11. `pop()`

Removes an arbitrary element. Do not depend on which element is removed.


```python
numbers = {10, 20, 30}
removed = numbers.pop()
print("Removed:", removed)
print("Remaining:", numbers)
```

## 12. `clear()`

```python
numbers = {10, 20, 30}
numbers.clear()
print(numbers)
```

## 13. `del`

`del` deletes the complete set variable.


```python
numbers = {10, 20, 30}
del numbers

# print(numbers)  # Uncommenting this gives NameError
```

## 14. Membership Operators — `in` and `not in`

```python
fruits = {"apple", "banana", "mango"}

if "mango" in fruits:
    print("Mango exists")

if "orange" not in fruits:
    print("Orange not found")
```

## 15. Union

Union means: **all unique elements from both sets**.

Methods:
- `A.union(B)`
- `A | B`


```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))
print(A | B)
```

## 16. Intersection

Intersection means: **common elements in both sets**.

Methods:
- `A.intersection(B)`
- `A & B`


```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.intersection(B))
print(A & B)
```

## 17. Difference

Difference means: **elements in the first set but not in the second set**.

Methods:
- `A.difference(B)`
- `A - B`

Remember: `A - B` and `B - A` are different.


```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("A - B:", A - B)
print("B - A:", B - A)
```

## 18. Symmetric Difference

Means: **elements present in either set, but not common in both**.

Methods:
- `A.symmetric_difference(B)`
- `A ^ B`


```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.symmetric_difference(B))
print(A ^ B)
```

## 19. Important Set Operation Comparison

Suppose:

`A = {1, 2, 3}`  
`B = {3, 4, 5}`

| Symbol | Meaning |
|---|---|
| `|` | All unique elements |
| `&` | Common elements |
| `-` | First set only |
| `^` | Non-common elements |


```python
A = {1, 2, 3}
B = {3, 4, 5}

print("Union A | B:", A | B)
print("Intersection A & B:", A & B)
print("Difference A - B:", A - B)
print("Symmetric Difference A ^ B:", A ^ B)
```

## 20. Subset

A set is a subset if all its elements exist in another set.

Methods:
- `A.issubset(B)`
- `A <= B`


```python
A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))
print(A <= B)
```

## 21. Superset

A set is a superset if it contains all elements of another set.

Methods:
- `B.issuperset(A)`
- `B >= A`


```python
A = {1, 2}
B = {1, 2, 3, 4}

print(B.issuperset(A))
print(B >= A)
```

## 22. Disjoint Sets

Two sets are disjoint if they have **no common elements**.


```python
A = {1, 2, 3}
B = {4, 5, 6}
print(A.isdisjoint(B))

C = {1, 2, 3}
D = {3, 4, 5}
print(C.isdisjoint(D))
```

## 23. Set Comprehension

General structure:

`{expression for variable in iterable}`


```python
# Normal way
squares = set()

for x in range(1, 6):
    squares.add(x * x)

print("Normal way:", squares)

# Set comprehension
squares2 = {x * x for x in range(1, 6)}
print("Comprehension:", squares2)
```

## 24. Set Comprehension with Condition

Pattern:

`{x for x in collection if condition}`


```python
numbers = {10, 20, 30, 40, 50}
result = {x for x in numbers if x > 25}
print(result)
```

## 25. Real-World Example — Remove Duplicate Student Names

```python
students = [
    "Ajinkya",
    "Rahul",
    "Ajinkya",
    "Sakshi",
    "Rahul"
]

unique_students = set(students)
print(unique_students)
```

## 26. Real-World Example — Common Skills

```python
ajinkya_skills = {
    "Python",
    "Java",
    "HTML",
    "CSS"
}

job_skills = {
    "Python",
    "Java",
    "SQL",
    "AI"
}

common = ajinkya_skills & job_skills
print("Common skills:", common)
```

## 27. Real-World Example — Missing Skills

```python
my_skills = {
    "Python",
    "Java",
    "HTML"
}

required_skills = {
    "Python",
    "Java",
    "SQL",
    "AI",
    "Git"
}

missing = required_skills - my_skills
print("Missing skills:", missing)
```

## 28. `frozenset`

A normal set is mutable (changeable).

A `frozenset` is immutable (locked / not changeable).


```python
numbers = frozenset({1, 2, 3})
print(numbers)

# numbers.add(4)  # Uncommenting this gives AttributeError
```

## 29. Important Restriction

A set cannot directly contain mutable objects like lists.

But it can contain tuples if their contents are hashable.


```python
data = {(1, 2), (3, 4)}
print(data)

# invalid_data = {[1, 2], [3, 4]}  # This gives TypeError
```
