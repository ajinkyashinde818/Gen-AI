# List

## A List is An ordered, mutable collection that can store multiple items of different data types.

### Different methods of list 
| Method      | Purpose                      |
| ----------- | ---------------------------- |
| `append()`  | Add one item                 |
| `extend()`  | Add multiple items           |
| `insert()`  | Insert at index              |
| `remove()`  | Remove by value              |
| `pop()`     | Remove by index              |
| `clear()`   | Remove all items             |
| `sort()`    | Sort list                    |
| `reverse()` | Reverse list                 |
| `copy()`    | Copy list                    |
| `count()`   | Count occurrences of a value |
| `index()`   | Find index of a value        |


```python
fruits = ["Apple", "Banana", "Mango"]
num=[1,2,3,4]
```

```python
num.append(5)
print(num)
```

```python
num.extend([6,7])
print(num)
```

```python
num.insert(2,7)
print(num)
```

```python
num.remove(3)
print(num)
```

```python
num.pop(2)
print(num)
```

```python
num.sort()
print(num)
```

```python
num.reverse()
print(num)
```

```python
num.count(5)
print(num)
```

```python
num.index(7,1)
print(num)
```

```python
''' 🛒 Real-World Problem: Grocery Store Billing System

You are developing software for a small grocery shop.

Requirements

Take input from the customer:

Customer Name
Mobile Number
5 grocery item names (store them in a list)
Price of each item
Quantity of each item
Customer Type (Regular or Premium)
Coupon Code '''

# ---------------- GROCERY BILLING SYSTEM ----------------

print("=========== GROCERY BILLING SYSTEM ===========")

# Customer Details
name = input("Enter Customer Name: ")
mobile = input("Enter Mobile Number: ")

customer_type = input("Customer Type (Regular/Premium): ")
coupon_code = input("Enter Coupon Code: ")

# Item Details
item1 = input("Enter Item 1 Name: ")
price1 = float(input("Enter Item 1 Price: "))
qty1 = int(input("Enter Item 1 Quantity: "))

item2 = input("Enter Item 2 Name: ")
price2 = float(input("Enter Item 2 Price: "))
qty2 = int(input("Enter Item 2 Quantity: "))

item3 = input("Enter Item 3 Name: ")
price3 = float(input("Enter Item 3 Price: "))
qty3 = int(input("Enter Item 3 Quantity: "))

item4 = input("Enter Item 4 Name: ")
price4 = float(input("Enter Item 4 Price: "))
qty4 = int(input("Enter Item 4 Quantity: "))

item5 = input("Enter Item 5 Name: ")
price5 = float(input("Enter Item 5 Price: "))
qty5 = int(input("Enter Item 5 Quantity: "))

# Store item names in a list
items = [item1, item2, item3, item4, item5]

# Calculate Total of Each Item
total1 = price1 * qty1
total2 = price2 * qty2
total3 = price3 * qty3
total4 = price4 * qty4
total5 = price5 * qty5

# Subtotal
subtotal = total1 + total2 + total3 + total4 + total5

# GST (5%)
gst = subtotal * 5 / 100

# Amount after GST
amount = subtotal + gst

# Premium Discount
discount = 0

if customer_type.lower() == "premium":
    discount = amount * 10 / 100

# Coupon
coupons = ["SAVE100", "WELCOME50", "DISCOUNT25"]

coupon_discount = 0

if coupon_code in coupons:
    coupon_discount = 100
else:
    print("Invalid Coupon")

# Delivery Charge
delivery = 0

if amount < 1000:
    delivery = 50

# Final Bill
final_bill = amount - discount - coupon_discount + delivery

# Payment
paid = float(input("Enter Amount Paid: "))

print("\n")

if paid < final_bill:
    print("Insufficient Payment")
else:
    return_amount = paid - final_bill

    print("==============================================")
    print("              GROCERY BILL")
    print("==============================================")

    print("Customer Name :", name)
    print("Mobile Number :", mobile)

    print("\nPurchased Items")
    print("----------------------------------------------")
    print(item1, "-", qty1, "x", price1, "=", total1)
    print(item2, "-", qty2, "x", price2, "=", total2)
    print(item3, "-", qty3, "x", price3, "=", total3)
    print(item4, "-", qty4, "x", price4, "=", total4)
    print(item5, "-", qty5, "x", price5, "=", total5)

    print("----------------------------------------------")
    print("Subtotal          :", subtotal)
    print("GST (5%)          :", gst)
    print("Premium Discount  :", discount)
    print("Coupon Discount   :", coupon_discount)
    print("Delivery Charge   :", delivery)
    print("----------------------------------------------")
    print("Final Bill        :", final_bill)
    print("Paid Amount       :", paid)
    print("Return Amount     :", return_amount)
    print("==============================================")
    print("Thank You! Visit Again.")
```
