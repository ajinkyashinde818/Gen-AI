# List

## A List is An ordered, mutable collection that can store multiple items of different data types.

### Different methods of list 
| Method      | Purpose                      |
| ----------- | ---------------------------- |
| `append()`  | Add one item                 |
| `extend()`  | Add multiple items           |
| `insert()`  | Insert at index              |
| `remove()`  | Remove by value              |
| `pop()`     | Remove by index              |
| `clear()`   | Remove all items             |
| `sort()`    | Sort list                    |
| `reverse()` | Reverse list                 |
| `copy()`    | Copy list                    |
| `count()`   | Count occurrences of a value |
| `index()`   | Find index of a value        |


```python
fruits = ["Apple", "Banana", "Mango"]
num=[1,2,3,4]
```

```python
num.append(5)
print(num)
```

```python
num.extend([6,7])
print(num)
```

```python
num.insert(2,7)
print(num)
```

```python
num.remove(3)
print(num)
```

```python
num.pop(2)
print(num)
```

```python
num.sort()
print(num)
```

```python
num.reverse()
print(num)
```

```python
num.count(5)
print(num)
```

```python
num.index(7,1)
print(num)
```
