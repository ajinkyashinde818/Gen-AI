# List

## A List is An ordered, mutable collection that can store multiple items of different data types.

### Different methods of list 
| Method      | Purpose                      |
| ----------- | ---------------------------- |
| `append()`  | Add one item                 |
| `extend()`  | Add multiple items           |
| `insert()`  | Insert at index              |
| `remove()`  | Remove by value              |
| `pop()`     | Remove by index              |
| `clear()`   | Remove all items             |
| `sort()`    | Sort list                    |
| `reverse()` | Reverse list                 |
| `copy()`    | Copy list                    |
| `count()`   | Count occurrences of a value |
| `index()`   | Find index of a value        |

## Enumerate -> A Enumerate fuctions is helps to to find both index and it's value while looping through the List 

```python
number=[32,45,34,23,65]

for index,value in enumerate(number):
    print(index,value)
```

```python
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits,start=1):
    print(index, fruit)
```

# List Comprehension

## We are going to use Looping statement inside the list so that we can store the values inside the list in an iterable format and in list comprehension conditional statement are allowed as well

```python
'without list comprehension'
numbers=[]

for i in range(5):
    numbers.append(i)

print(numbers)
```

```python
'with list comprehension'
list_items=[x for x in range(1,51)]
print(list_items)
```

```python
mult=[i*5 for i in range(50)]
print(mult)
```

```python
# Odd Number 

odd=[i for i in range(1,101) if i % 2 == 0]
print(odd)


# Even Number

even=[i for i in range(1,101) if i % 2 != 0]
print(even)
```
