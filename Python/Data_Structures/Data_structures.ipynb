# Data Structures in Python 

### A Data Structures in Python is a way to Storing and organizing a data in a efficient manner and easy accessible and perform a basic operations such as CRUD operations is called as the Data Structures 

## Different types of data structures

| Data Structure | Ordered | Mutable | Allows Duplicates         | Example              |
| -------------- | ------- | ------- | ------------------------- | -------------------- |
| List           | ✅ Yes   | ✅ Yes   | ✅ Yes                     | `[10, 20, 30]`       |
| Tuple          | ✅ Yes   | ❌ No    | ✅ Yes                     | `(10, 20, 30)`       |
| Set            | ❌ No    | ✅ Yes   | ❌ No                      | `{10, 20, 30}`       |
| Dictionary     | ✅ Yes*  | ✅ Yes   | Keys: ❌ No, Values: ✅ Yes | `{"name":"Ajinkya"}` |


```python
fruit=["Apple","Mango","Pinapple"]
print(fruit)
print(type(fruit))
```

```python
car=("Swift","Fortuner","Ferreri","GT 650")
print(car)
print(type(car))
```

```python
age={12,45,32,46}
print(age)
print(type(age))
```

```python
info={
    "name":"Ajinkya",
    "Age":"18",
    "branch":"Computer Technology"
}

print(info)
print(type(info))
```
