# Looping Statement: `while` Loop and Loop Control Statements

This notebook explains the Python `while` loop and loop control statements (`break`, `continue`, and `pass`) with simple examples.

## 1. While Loop
A **while loop** repeats a block of code as long as the condition is **True**.

**Syntax:**
```python
while condition:
    # code
```

````python
i = 1

while i <= 5:
    print(i)
    i += 1
````

### Countdown Example

````python
num = 5

while num > 0:
    print(num)
    num -= 1

print("Blast Off!")
````

### Password Example

````python
password = ""

while password != "python":
    password = input("Enter Password: ")

print("Access Granted")
````

## 2. Loop Control Statements

### A. `break`
Stops the loop immediately.

````python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
````

### B. `continue`
Skips the current iteration and moves to the next one.

````python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````

### C. `pass`
Does nothing. Used as a placeholder.

````python
for i in range(5):
    if i == 2:
        pass
    print(i)
````

## Real-World Example: ATM PIN Check

````python
correct_pin = "1234"

while True:
    pin = input("Enter PIN: ")

    if pin == correct_pin:
        print("Login Successful")
        break

    print("Wrong PIN")
````

## Real-World Example: Print Even Numbers

````python
i = 1

while i <= 10:
    if i % 2 != 0:
        i += 1
        continue

    print(i)
    i += 1
````

## Difference Between `break`, `continue`, and `pass`

| Statement | Purpose |
|-----------|---------|
| `break` | Stops the loop completely |
| `continue` | Skips the current iteration |
| `pass` | Does nothing (placeholder) |
