#  Looping Statements - `for` Loop 

This notebook explains only the **`for` loop** with simple explanations and examples.


## 1. What is a `for` Loop?

Use a `for` loop when you want to repeat a task multiple times.


```python
for i in range(5):
    print('Hello')
```

## 2. Syntax

General syntax of a `for` loop.


```python
#for variable in sequence:
#    print(variable)
```

## 3. `range(stop)`

Starts from 0 and stops before the given number.


```python
for i in range(5):
    print(i)
```

## 4. `range(start, stop)`

Starts from the first number and stops before the second.


```python
for i in range(2,6):
    print(i)
```

## 5. `range(start, stop, step)`

Third value is the step size.


```python
for i in range(2,11,2):
    print(i)
```

## 6. Negative Step

Count backwards.


```python
for i in range(10,0,-1):
    print(i)
```

## 7. Loop Through a String

Iterate over every character.


```python
name='Ajinkya'
for ch in name:
    print(ch)
```

## 8. Loop Through a List

Iterate over list elements.


```python
fruits=['Apple','Banana','Mango']
for fruit in fruits:
    print(fruit)
```

## 9. Loop Through a Tuple

Iterate over tuple elements.


```python
numbers=(10,20,30)
for num in numbers:
    print(num)
```

## 10. Loop Through a Set

Sets are unordered.


```python
colors={'Red','Green','Blue'}
for color in colors:
    print(color)
```

## 11. Loop Through a Dictionary

Using items() returns keys and values.


```python
student={'name':'Ajinkya','marks':91}
for key,value in student.items():
    print(key,':',value)
```

## 12. Using len()

Access index and value together.


```python
fruits=['Apple','Banana','Mango']
for i in range(len(fruits)):
    print(i, fruits[i])
```

## 13. Multiplication Table

Real-world example.


```python
num=7
for i in range(1,11):
    print(f'{num} x {i} = {num*i}')
```

## 14. Sum of Numbers

Add numbers using a loop.


```python
total=0
for i in range(1,6):
    total+=i
print(total)
```

## 15. Factorial

Find factorial using a loop.


```python
num=5
fact=1
for i in range(1,num+1):
    fact*=i
print(fact)
```

## 16. Nested for Loop

Loop inside another loop.


```python
for i in range(1,4):
    for j in range(1,4):
        print(i,j)
```

# Common Mistakes

- Forgetting the colon (`:`)
- Wrong indentation
- Thinking `range(5)` includes 5 (it does not)
- Using the wrong step value


# Quick Revision

- `range(5)` → 0 to 4
- `range(1,6)` → 1 to 5
- `range(2,11,2)` → 2,4,6,8,10
- `range(10,0,-1)` → Reverse counting
- Loop through strings, lists, tuples, sets, and dictionaries.


```python
'''1. Print Numbers

Write a program to print numbers from 1 to 10 using a for loop.

2. Multiplication Table

Take a number from the user and print its multiplication table from 1 to 10.

3. Sum of Numbers

Take a number N from the user and find the sum of numbers from 1 to N.

4. Even Numbers

Print all even numbers from 1 to 50 using a for loop.

5. Count Vowels

Take a string from the user and count the total number of vowels (a, e, i, o, u) using a for loop.
'''
```

```python
# 1.
for i in range(1,11):
    print(i)
```

```python
# 2.

num=int(input("Enter a Number:"))

for i in range(1,11):
    multiply=num*i
    print(num,"*",i,"=",multiply)
```

```python
# 3.

num=int(input("Enter a number:"))


total=0
for i in range(1,num+1):
    total+=i
print("The Sum of Numbers:",total)
```

```python
# 4.
for i in range(0,51):
    if i% 2==0:
        print("Even Numbers:",i)
```

```python
# 5.

name=input("Enter a String:")
convert=name.lower()

count=0

vowel=['a','e','i','o','u']

for i in convert:
    if i in vowel:
        count=count+1

print("Counted Vowels:",count)
```
