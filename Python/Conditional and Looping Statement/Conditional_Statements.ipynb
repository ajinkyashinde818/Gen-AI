## 1. Basic `if` Statement

Use `if` when you want to execute code only when a condition is `True`.

**Syntax:**
```python
if condition:
    statement
```

Python uses indentation (spaces) to identify code inside the `if` block.


````python
age = 20

if age >= 18:
    print("You are eligible to vote")
````

## 2. `if-else` Statement

Use `if-else` when there are two possibilities:
- condition is `True`
- condition is `False`


````python
age = 16

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
````

## 3. `if-elif-else`

Use this when there are multiple conditions.

Python checks conditions from **top to bottom** and stops at the first `True` condition.


````python
marks = 75

if marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
````

## 4. Comparison Operators

| Operator | Meaning | Example |
|---|---|---|
| `==` | Equal to | `age == 18` |
| `!=` | Not equal to | `age != 18` |
| `>` | Greater than | `age > 18` |
| `<` | Less than | `age < 18` |
| `>=` | Greater than or equal | `age >= 18` |
| `<=` | Less than or equal | `age <= 18` |

### Important
- `=` means assignment
- `==` means comparison


````python
password = "admin123"

if password == "admin123":
    print("Login successful")
else:
    print("Wrong password")
````

## 5. Logical Operator `and`

`and` means **all conditions must be True**.

| Condition 1 | Condition 2 | Result |
|---|---|---|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |


````python
percentage = 91
documents = True

if percentage >= 90 and documents == True:
    print("Admission eligible")
else:
    print("Not eligible")
````

## 6. Logical Operator `or`

`or` means **at least one condition must be True**.


````python
is_admin = False
is_teacher = True

if is_admin or is_teacher:
    print("Access granted")
else:
    print("Access denied")
````

## 7. Logical Operator `not`

`not` reverses a Boolean result:
- `not True` → `False`
- `not False` → `True`


````python
is_blocked = False

if not is_blocked:
    print("User can login")
else:
    print("User is blocked")
````

## 8. Nested `if`

A nested `if` means an `if` statement inside another `if` statement.


````python
age = 20
has_license = True

if age >= 18:
    print("Age requirement passed")

    if has_license:
        print("You can drive")
    else:
        print("License required")
else:
    print("You are underage")
````

## 9. Real-World Login System

Both username and password must be correct, so we use `and`.


````python
username = input("Enter username: ")
password = input("Enter password: ")

if username == "ajinkya" and password == "1234":
    print("Login successful")
else:
    print("Invalid username or password")
````

## 10. Student Grade System with Validation

First validate that marks are between `0` and `100`.
Then check grades from highest to lowest.


````python
marks = int(input("Enter marks: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 40:
    print("Pass")
else:
    print("Fail")
````

## 11. Positive, Negative, or Zero


````python
num = int(input("Enter number: "))

if num > 0:
    print("Positive number")
elif num < 0:
    print("Negative number")
else:
    print("Zero")
````

## 12. Even or Odd

The `%` operator gives the remainder.

If a number divided by `2` gives remainder `0`, it is even.


````python
num = int(input("Enter number: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
````

## 13. Membership Condition with `in`

Use `in` to check whether a value exists inside a list, set, tuple, string, or dictionary.


````python
students = ["Ajinkya", "Rahul", "Sakshi"]

name = input("Enter student name: ")

if name in students:
    print("Student found")
else:
    print("Student not found")
````

## 14. Using `not in`

Use `not in` when you want to check that a value does not exist in a collection.


````python
blocked_users = ["Rahul", "Om"]

name = input("Enter name: ")

if name not in blocked_users:
    print("Access allowed")
else:
    print("User is blocked")
````

## 15. Conditional Statement with Dictionary


````python
student = {
    "name": "Ajinkya",
    "marks": 91
}

if student["marks"] >= 90:
    print("Excellent student")
else:
    print("Keep improving")
````

## 16. Conditional Statement with List


````python
marks = [91, 85, 78, 95]

if 95 in marks:
    print("95 marks found")
else:
    print("Not found")
````

## 17. Short-Hand `if`

A one-line `if` is valid for very simple statements.


````python
age = 20

if age >= 18: print("Adult")
````

## 18. Ternary Conditional Expression

Normal form:
```python
if condition:
    value = "something"
else:
    value = "other"
```

Short form:
```python
value = true_value if condition else false_value
```


````python
age = 20

status = "Adult" if age >= 18 else "Minor"

print(status)
````

## 19. Truthy and Falsy Values

Common **Falsy** values:
- `False`
- `None`
- `0`
- `0.0`
- `""`
- `[]`
- `{}`
- `set()`

Other non-empty values are generally Truthy.


````python
name = ""

if name:
    print("Name exists")
else:
    print("Name is empty")
````

````python
students = []

if students:
    print("Students available")
else:
    print("No students available")
````

## 20. Common Mistakes

### Mistake 1: Using `=` instead of `==`
Wrong:
```python
if age = 18:
```

Correct:
```python
if age == 18:
```

### Mistake 2: Missing colon `:`
Wrong:
```python
if age >= 18
```

Correct:
```python
if age >= 18:
```

### Mistake 3: Wrong indentation
Code inside an `if` block must be indented.

### Mistake 4: Wrong condition order
When checking marks, check higher ranges first.


## 21. Complete Real-World Example — Scholarship Eligibility System


````python
name = input("Enter student name: ")
percentage = float(input("Enter percentage: "))
income = int(input("Enter family income: "))
category = input("Enter category: ")

if percentage < 0 or percentage > 100:
    print("Invalid percentage")

elif percentage >= 90 and income <= 300000:
    print(name, "is eligible for Merit Scholarship")

elif category == "SC" and percentage >= 75:
    print(name, "is eligible for Category Scholarship")

elif percentage >= 60:
    print(name, "is eligible for General Support")

else:
    print(name, "is not eligible for scholarship")
````
