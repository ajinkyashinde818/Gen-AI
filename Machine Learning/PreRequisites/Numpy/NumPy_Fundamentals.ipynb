# 🔢 NumPy — Introduction

**NumPy = Numerical Python**

NumPy is an important Python library for **Data Science, Machine Learning, numerical calculations, arrays, matrices, statistics, and scientific computing**.

The main object in NumPy is the **NumPy Array (`ndarray`)**.

## 1. Why NumPy?

Python lists are general-purpose collections. NumPy arrays are designed for fast numerical operations.

```python
numbers = [10, 20, 30, 40]
print(numbers * 2)
```

Output:

```text
[10, 20, 30, 40, 10, 20, 30, 40]
```

With NumPy:

```python
import numpy as np

arr = np.array([10, 20, 30, 40])
print(arr * 2)
```

Output:

```text
[20 40 60 80]
```

NumPy performs the operation on every element.

## 4. Creating a NumPy Array

Use `np.array()`:

```python
arr = np.array([10, 20, 30, 40])
print(arr)
```

Output:

```text
[10 20 30 40]
```

## 5. 1-D, 2-D and 3-D Arrays

### 1-D

```python
np.array([10, 20, 30, 40])
```

### 2-D

```python
np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

Visual:

```text
1  2  3
4  5  6
```

### 3-D

```python
np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
```

Think of a 3-D array as **multiple 2-D matrices**.

## 6. Important Array Properties

```python
arr.ndim
arr.shape
arr.size
arr.dtype
```

| Property | Meaning |
|---|---|
| `ndim` | Number of dimensions |
| `shape` | Size of each dimension |
| `size` | Total number of elements |
| `dtype` | Data type |

````python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data Type:", arr.dtype)
````

## 7. Indexing

NumPy uses zero-based indexing.

```python
arr = np.array([10, 20, 30, 40])
```

```text
Value:    10   20   30   40
Index:     0    1    2    3
```

Examples:

```python
arr[0]   # 10
arr[2]   # 30
```

### Negative indexing

```python
arr[-1]  # 40
arr[-2]  # 30
```

## 8. 2-D Array Indexing

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
```

To access `50`:

```python
arr[1, 1]
```

General form:

```python
arr[row, column]
```

## 9. Slicing

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])
```

Output:

```text
[20 30 40]
```

General form:

```text
[start : stop]
```

The `stop` index is not included.

## 10. NumPy Arithmetic

```python
arr = np.array([10, 20, 30, 40])
```

```python
arr + 5    # [15 25 35 45]
arr - 5    # [ 5 15 25 35]
arr * 2    # [20 40 60 80]
arr / 2    # [ 5. 10. 15. 20.]
```

````python
arr = np.array([10, 20, 30, 40])

print("Addition:", arr + 5)
print("Subtraction:", arr - 5)
print("Multiplication:", arr * 2)
print("Division:", arr / 2)
````

## 11. Array + Array

Operations can be performed element by element.

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

a + b   # [11 22 33]
a - b   # [ 9 18 27]
a * b   # [10 40 90]
a / b   # [10. 10. 10.]
```

## 12. Mathematical Functions

Important functions:

```python
np.sum(arr)
np.min(arr)
np.max(arr)
np.mean(arr)
np.std(arr)
```

````python
arr = np.array([10, 20, 30, 40])

print("Sum:", np.sum(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Mean:", np.mean(arr))
print("Standard Deviation:", np.std(arr))
````

## 13. Special Array Creation

### `np.zeros()`

```python
np.zeros(5)
```

### `np.ones()`

```python
np.ones(5)
```

### `np.arange()`

```python
np.arange(1, 10)
np.arange(1, 10, 2)
```

### `np.linspace()`

```python
np.linspace(0, 10, 5)
```

````python
print("Zeros:", np.zeros(5))
print("Ones:", np.ones(5))
print("Arange:", np.arange(1, 10))
print("Arange with step:", np.arange(1, 10, 2))
print("Linspace:", np.linspace(0, 10, 5))
````

## 14. Random Numbers

### Random decimals

```python
np.random.rand(5)
```

Creates 5 random decimal values between 0 and 1.

### Random integers

```python
np.random.randint(1, 100, 5)
```

Creates 5 random integers from 1 to 99.

````python
print("Random decimals:", np.random.rand(5))
print("Random integers:", np.random.randint(1, 100, 5))
````
