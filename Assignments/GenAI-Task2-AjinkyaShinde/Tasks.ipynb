# Task 1: Discount Rules (if / elif / else)

```python
# 1.Write a program that reads an integer order_amount from the user using input().

order_amount=int(input("Enter a Order Amount"))
print("A Order Amount is:",order_amount)
```

```python
# 2.Apply the following discount rules and print the final amount:
'''order_amount >= 2000 → 15% discount
   1500 <= order_amount < 2000 → 10% discount
   1000 <= order_amount < 1500 → 7% discount
   otherwise → 0% discount '''

sub_total=0
if order_amount >=2000:
    discount=order_amount*15/100
    sub_total=order_amount-discount

elif order_amount>=1500:
    discount=order_amount*10/100
    sub_total=order_amount-discount

elif order_amount>=1000:
    discount=order_amount*7/100
    sub_total=order_amount-discount

else:
    discount=0
    sub_total=order_amount-discount


print("A Order Amount is:",order_amount)
print("A Discount:",discount)
print("A Final Amount:",sub_total)
```

```python
# Ensure you convert input to a number and handle the case when the input is non-numeric by displaying an error message and exiting the program.

user=input("Enter a Value:")

if user.isdigit():
    print("Converted to integer:",user)
else:
    print("Error....:")
```
